```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
from datetime import date, timedelta
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://queenannessheriff.org/most-wanted/"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    list1 = driver.find_elements(By.XPATH, f'/html/body/div/div/div[3]/div')
    
    for i in range (1, len(list1)):
        data_dict = {}
        driver.find_element(By.XPATH, f'/html/body/div/div/div[3]/div[{i}]/h2/a').click()
        fullName = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/h2').text
        name = fullName.split(",")
        firstName = name[1]
        middleName = firstName.split(" ")[1]
        firstName = firstName.split(" ")[0]
        lastName = name[0]
        li = len(driver.find_elements(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li'))
        for k in range (1, li+1):
            headings = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]/strong').text
            if "Offense:"in headings:
                offence = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
                print(offence)
            elif "Date Entered:" in headings:
                dateEntered = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
                print(dateEntered)
            elif "Last Known Address:" in headings:
                address = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
                print(address)
            elif "Birthdate:" in headings:
                dob = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
                print(dob)
            elif "Height:" in headings:
                height = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].replace('\"', '').strip()
                print(height)
            elif "Weight:" in headings:
                weight = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
                print(weight)
            elif "Race:" in headings:
                race = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
            elif "Gender:" in headings:
                gender = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
            elif "Eye Color:" in headings:
                eyes = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
            elif "Hair Color:" in headings:
                hair = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div/ul/li[{k}]').text.split(":", 1)[1].strip()
        importantDates = "Date Entered: " + dateEntered
        listOfAddress = "Last Known Address: " + address
        summary = fullName + " is wanted for the offence " + offence + " has " + eyes + " eyes, and " + hair + "hairs." 
        image = ""
        try:
            image = driver.find_element(By.XPATH, f'/html/body/div/div/div[2]/div[1]/div/div[1]/img').get_attribute("src")
        except:
            pass
        driver.back()
        if fullName:
            data_dict['fullName'] = fullName
        if firstName:
            data_dict['firstName'] = firstName
        if middleName:
            data_dict['middleName'] = middleName
        if lastName:
            data_dict['lastName'] = lastName
        if offence:
            data_dict['offence'] = offence
        if image:
            data_dict['image'] = image
        if gender:
            data_dict['gender'] = gender
        if race:
            data_dict['race'] = race
        if dob:
            data_dict['dob'] = dob
        if height:
            data_dict['height'] = height
        if weight:
            data_dict['weight'] = weight
        if hair:
            data_dict['hair'] = hair
        if eyes:
            data_dict['eyes'] = eyes
        if importantDates:
            data_dict['importantDates'] = importantDates
        if listOfAddress:
            data_dict['listOfAddress'] = listOfAddress
        if summary:
            data_dict['summary'] = summary
        data_list.append(data_dict)
    return data_list
```

```python
if __name__ == '__main__':
    data_list = get_data("add_slug_name")
    to_json(data_list)
```
