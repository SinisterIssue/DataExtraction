```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
import re
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://www.oakgov.com/sheriff/Law-Enforcement/Pages/OCSO-Most-Wanted-Fugitives.aspx"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    list1 = driver.find_elements(By.XPATH, f'/html/body/form/div[4]/div/div[2]/div[2]/div/main/div[2]/section/div/div/div/div/div[1]/div/oc-app/app-view/oc-gallery-masonry/div[1]/oc-masonry/div/div/div')
    
    for i in range (2, len(list1)+2):
        data_dict={}
        driver.find_element(By.XPATH, f'/html/body/form/div[4]/div/div[2]/div[2]/div/main/div[2]/section/div/div/div/div/div[1]/div/oc-app/app-view/oc-gallery-masonry/div[1]/oc-masonry/div/div[{i}]/div').click()
        fullName = driver.find_element(By.XPATH, f'/html/body/form/div[4]/div/div[2]/div[2]/div/main/div[2]/section/div/div/div/div/div[1]/div/oc-app/app-view/oc-gallery-masonry/div[2]/div[2]/div[2]/div[4]/div/h2').text
        if(len(fullName.split(" ")) ==2):
            firstName = fullName.split(" ")[0]
            lastName = fullName.split(" ")[1]
            middleName = ""
        if(len(fullName.split(" ")) ==3):
            firstName = fullName.split(" ")[0]
            middleName = fullName.split(" ")[1]
            lastName = fullName.split(" ")[2]
        print(fullName)
        time.sleep(5)
        driver.find_element(By.XPATH, f'/html/body/form/div[4]/div/div[2]/div[2]/div/main/div[2]/section/div/div/div/div/div[1]/div/oc-app/app-view/oc-gallery-masonry/div[2]/div[2]/div[2]/div[1]/a[1]').click()
        image = driver.find_element(By.XPATH, f'/html/body/form/div[4]/div/div[2]/div[2]/div/main/div[2]/section/div/div/div/div/div[1]/div/oc-app/app-view/oc-gallery-masonry/div[2]/div[2]/div[1]/div[2]/div/img[2]').get_attribute("src")
        print(image)
        additionalInfo = driver.find_element(By.XPATH, f'/html/body/form/div[4]/div/div[2]/div[2]/div/main/div[2]/section/div/div/div/div/div[1]/div/oc-app/app-view/oc-gallery-masonry/div[2]/div[2]/div[2]/div[4]/div/p').text
        additionalInfo = additionalInfo.replace("\u2019", "\'").replace('\u201c', '').replace("\u201d", '').replace("\u2013", "-")
        summary = fullName + "  is wanted by the Oakland County Sheriff's Office FOC Unit."
        info = re.sub("DOB:", ";DOB:", additionalInfo)
        info = re.sub("Eyes:", ";Eyes:", info)
        info = re.sub("Hair:", ";Hair:", info)
        info = re.sub("Height:", ";Height:", info)
        info = re.sub("Weight:", ";Weight:", info)
        info = re.sub("Birth Date:", ";Birth Date:", info)
        info = re.sub(firstName, ";"+firstName, info)
#         print(info)
        info = info.split(";")
#         print(info)
#         print(len(info))
        eyes = ""
        hair = ""
        for j in range(1, len(info)+1):
            try:
                if "DOB:" in info[j]:
                    dob = info[j].split(":")[1].strip()
                    d = dob.split("/")[1]
                    m = dob.split("/")[0]
                    y = dob.split("/")[2]
                    dob_dmy = d + "/" + m + "/" + y
                    print(dob_dmy)
                elif "Birth Date:" in info[j]:
                    dob = info[j].split(":")[1].strip()
                    d = dob.split("/")[1]
                    m = dob.split("/")[0]
                    y = dob.split("/")[2]
                    dob_dmy = d + "/" + m + "/" + y
                    print(dob_dmy)
                elif "Height:" in info[j]:
                    height = info[j].split(":")[1].strip()
                    print(height)
                elif "Weight:" in info[j]:
                    weight = info[j].split(":")[1].strip()
                    print(weight)
                elif "Eyes:" in info[j]:
                    eyes = info[j].split(":")[1].strip()
                    print(eyes)
                elif "Hair:" in info[j]:
                    hair = info[j].split(":")[1].strip()
                    print(hair)
            except:
                pass
                    
#         print(additionalInfo)
        
        if fullName:
            data_dict['fullName'] = fullName
            if firstName:
                data_dict['firstName'] = firstName
            if middleName:
                data_dict['middleName'] = middleName
            if lastName:
                data_dict['lastName'] = lastName
            if image:
                data_dict['image'] = image
            if dob_dmy:
                data_dict['dob'] = dob_dmy
            if height:
                data_dict['height'] = height
            if weight:
                data_dict['weight'] = weight
            if eyes:
                data_dict['eyes'] = eyes
            if hair:
                data_dict['hair'] = hair
            if additionalInfo:
                data_dict['additionalInfo'] = additionalInfo
            if summary:
                data_dict['summary'] = summary
        data_list.append(data_dict)
    return data_list
```

```python
if __name__ == '__main__':
    data_list = get_data("add_slug_name")
    to_json(data_list)
```
