```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
import re
from deep_translator import GoogleTranslator
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://www.bundestag.de/en/members/"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
translator = GoogleTranslator(target='english')
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    driver.find_element(By.XPATH, f'/html/body/main/section[1]/div[2]/header/div[2]/div/form/a/i[1]').click()
    time.sleep(2)
    list1 = driver.find_elements(By.XPATH, f'/html/body/main/section[1]/div[2]/div/div[3]/div/ul/li/a/div/div/h3')
    for i in range(1, len(list1)+1):
        data_dict = {}
        maritalStatus = ""
        familyInfo = ""
        religion = ""
        careerInfo = ""
        placeOfBirthCountry =""
        placeOfBirthCity=""
        name = driver.find_element(By.XPATH, f'/html/body/main/section[1]/div[2]/div/div[3]/div/ul/li[{i}]/a/div/div/h3').text
        firstName = name.split(",")[1].strip()
        lastName = name.split(",")[0]
        fullName = firstName + " " + lastName
        print(fullName)
        politicalParty = driver.find_element(By.XPATH, f'/html/body/main/section[1]/div[2]/div/div[3]/div/ul/li[{i}]/a/div/div/p').text
    #     print(politicalParty)
        link = driver.find_element(By.XPATH, f'/html/body/main/section[1]/div[2]/div/div[3]/div/ul/li[{i}]/a').get_attribute("href")
        driver.execute_script("window.open('');")
        driver.switch_to.window(driver.window_handles[1])
        driver.get(link)
        time.sleep(2)
        careerInfoDesignation = driver.find_element(By.XPATH, f'/html/body/main/div[4]/div/div/div[1]/div[2]/div/p').text
        careerInfoDesignation = translator.translate(careerInfoDesignation)
    #     print(careerInfoDesignation)
        image = driver.find_element(By.XPATH, f'/html/body/main/div[4]/div/div/div[1]/div[1]/div/img').get_attribute('src')
    #     image = "https://www.bundestag.de" + image
        print(image)
        try:
            fullAddress = driver.find_element(By.XPATH, f'/html/body/main/div[4]/div/div/div[1]/div[5]/div/div[1]/p[1]').text.replace("\n", ", ")
    #     print(fullAddress)
        except:
            pass
        info = driver.find_element(By.XPATH, f'/html/body/main/div[4]/div/div/div[2]/div[1]/div/div[2]/div[2]/div').text
        info = translator.translate(info)
        info =info.replace("[Editor's note: The biographical information is based on the self-declaration of the MPs.]", "")
        dobInfo = info.split("\n")[0]
        careerInfo = info.split("\n", 1)[1]
        careerInfo = careerInfo.replace("\n", "; ").strip().replace("; ;", "").replace(";;", ";")
    #     print(dobInfo)
        if ";" not in dobInfo:
            try:
                placeOfBirthCity= dobInfo.split("in",1)[1].strip()
            except:
                pass
            dob= dobInfo.split("in", 1)[0].replace("Born", "").replace("on", "", 1).strip()
            try:
                placeOfBirthCountry = placeOfBirthCity.split(",")[1].strip()
                placeOfBirthCity = placeOfBirthCity.split(",")[0].strip()
                if placeOfBirthCountry =="married" or placeOfBirthCountry=="single"or placeOfBirthCountry=="divorced":
                    maritalStatus = placeOfBirthCountry
                    placeOfBirthCountry = ""
            except:
                pass
        if ";" in dobInfo:
            dob= dobInfo.split(";")[0]
            try:
                placeOfBirthCity= dob.split("in",1)[1]
            except:
                pass
            try:
                placeOfBirthCountry = placeOfBirthCity.split(",")[1].strip()
                placeOfBirthCity = placeOfBirthCity.split(",")[0].strip()
            except:
                pass
            dob = dob.split("in", 1)[0].replace("Born", "").replace("on", "", 1).strip()
        
        if ";" not in dobInfo and len(dob.split(","))>=3:
            
            try:
                dobInfo = dobInfo.split(",")
                print(dobInfo)
                for j in range (0, len(dobInfo)):
                    if "married" in dobInfo[j] or "single" in dobInfo[j] or "divorced" in dobInfo[j]:
                        maritalStatus = dobInfo[j]

                    elif "children" in dobInfo[j] or "child" in dobInfo[j]:
                        familyInfo = dobInfo[j]
                    elif "Born" in dobInfo[j]:
                        dobDay = dobInfo[j]
                    elif "in" in dobInfo[j]:
                        dobYear = dobInfo[j]
                        placeOfBirthCity = dobYear.split("in")[1].strip()
                        dobYear = dobYear.split("in")[0].strip
                dob = dobDay + " " + dobYear
            except:
                try:
                    info2 = dobInfo.split(".")[1]
                    careerInfo = info2 + " " + careerInfo
                    dobInfo = dobInfo.split(".")[0]
                    dobInfo = dobInfo.split(",")
                    print(dobInfo)
                    for j in range (0, len(dobInfo)):
                        if "married" in dobInfo[j] or "single" in dobInfo[j] or "divorced" in dobInfo[j]:
                            maritalStatus = dobInfo[j]
                        if "Born" in dobInfo[j]:
                            dobDay = dobInfo[j]
                        dobYear = dobInfo[1]
                        religion = dobYear.split(" ")[1]
                        dobYear = dobYear.split(" ")[0]
                    dob = dobDay + " " + dobYear
                except:
                    pass
                    
        
        elif ";" in dobInfo:
            dobInfo = dobInfo.split(";")
            print(dobInfo)
            for j in range (0, len(dobInfo)):
                if "married" in dobInfo[j] or "single" in dobInfo[j] or "divorced" in dobInfo[j]:
                    maritalStatus = dobInfo[j]
                    if "and" in maritalStatus:
                        religion = maritalStatus.split("and")[1].strip()
                        maritalStatus = maritalStatus.split("and")[0].strip()

                elif "children" in dobInfo[j] or "child" in dobInfo[j]:
                    familyInfo = dobInfo[j]
            try:
                if familyInfo:
                    dobInfo.remove(familyInfo)
                if maritalStatus:
                    dobInfo.remove(maritalStatus)
                if "Born" in dobInfo[0]:
                    dobInfo.pop(0)
                if len(dobInfo) >0:
                    religion = dobInfo[0]
            except:
                pass
        elif ";" in dobInfo and len(dobInfo.split(";"))>=2 and len(dobInfo.split(";")[0].split(","))>=3:
            familyInfo = dobInfo.split(";")[1]
            dobInfo = dobInfo.split(";")[0].split(",")
            for j in range (0, len(dobInfo)):
                if "married" in dobInfo[j] or "single" in dobInfo[j] or "divorced" in dobInfo[j]:
                        maritalStatus = dobInfo[j]

                elif "children" in dobInfo[j] or "child" in dobInfo[j]:
                    familyInfo = dobInfo[j]
                elif "Born" in dobInfo[j]:
                    dobDay = dobInfo[j]
                elif "in" in dobInfo[j]:
                    dobYear = dobInfo[j]
                    placeOfBirthCity = dobYear.split("in")[1].strip()
                    dobYear = dobYear.split("in")[0].strip
            dob = dobDay + " " + dobYear
            if familyInfo:
                dobInfo.remove(familyInfo)
            if maritalStatus:
                dobInfo.remove(maritalStatus)
            if "Born" in dobInfo[0]:
                dobInfo.pop(0)
            if len(dobInfo) >0:
                religion = dobInfo[0]
        if "Berghegger" in fullName:
            religion = "Roman Catholic"
        if "partnered" in religion:
            religion = ""
        if "children" in religion:
            familyInfo = familyInfo + ", "+ religion
            religion = ""
        
        
        dob = dob.split("in", 1)[0].replace("Born", "").replace("on", "", 1).strip()
        familyInfo = familyInfo.replace(".", "").strip()
        maritalStatus = maritalStatus.replace(".", "").strip()
        religion = religion.replace(".", "").strip()
        print(familyInfo)
        print(maritalStatus)
        print(religion)
    #     print(dob)
    #     print(placeOfBirthCity)
        print(dobInfo)
        print(careerInfo)
        summary = fullName + " is the " + careerInfoDesignation + " THE REPUBLIC OF BULGARIA " + "and belongs to the political party " + politicalParty
        print("*"*50)
        driver.close()
        driver.switch_to.window(driver.window_handles[0])
        if fullName:
            data_dict['fullName'] = fullName
        if firstName:
            data_dict['firstName'] = firstName
        if lastName:
            data_dict['lastName'] = lastName
        if image:
            data_dict['image'] = image
        if careerInfoDesignation:
            data_dict['careerInfoDesignation'] = careerInfoDesignation
        if politicalParty:
            data_dict['politicalParty'] = politicalParty
        if fullAddress:
            data_dict['fullAddress'] = fullAddress
        if dob:
            data_dict['dob'] = dob
        if placeOfBirthCity:
            data_dict['placeOfBirthCity'] = placeOfBirthCity
        if placeOfBirthCountry:
            data_dict['placeOfBirthCountry'] = placeOfBirthCountry
        if maritalStatus:
            data_dict['maritalStatus'] = maritalStatus
        if religion:
            data_dict['religion'] = religion
        if familyInfo:
            data_dict['familyInfo'] = familyInfo
        if careerInfo:
            data_dict['careerInfo'] = careerInfo
        if summary:
            data_dict['summary'] = summary
        data_list.append(data_dict)
    return data_list
```
