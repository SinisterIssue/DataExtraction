```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://www.woodsonsheriff.com/roster.php?released=1"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    data_dict = {}
    driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div[3]/div/div/div[2]/div[7]/div/em/a').click()
    div = len(driver.find_elements(By.XPATH, f'//div[1]/div/div[2]/div/div[@class="inmate_profile_data_bold cell"]/span'))
    for i in range (1, div+1):
        try:
            headings = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div/div[1]/div/div[2]/div[{i}]/div[1]').text
#             print(headings)
            if "Booking #:"in headings:
                bookId = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
                print(bookId)
            elif "Age:" in headings:
                age = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
                print(age)
            elif "Gender:" in headings:
                gender = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
                print(gender)
            elif "Race:" in headings:
                race = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
                print(race)
            elif "Arresting Agency:" in headings:
                agency = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
                print(agency)
            elif "Booking Date:" in headings:
                bookDate = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
                print(bookDate)
            elif "Release Date:" in headings:
                releaseDate = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{i}]/div[@class="cell inmate_profile_data_content"]').text
        except:
            pass
    arrestWarrant = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div/div[1]/div/div[2]/div[9]/div/span').text.replace("\n", "")
#     print(charges)
    bond = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div/div[1]/div/div[2]/div[10]/div[2]').text
#     print(bond)
    info = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div/div[2]/div').text
    additionalInfo = "Booking #:" + bookId + ", " + "Arresting Agency:" + agency + ", " + "Bond:" + bond + "," + info 
#     print(additionalInfo)
    importantDates = "Booking Date:" + bookDate + ", " + "Release Date:" + releaseDate 
    image = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div/div[1]/div/div[1]/img').get_attribute("src")
    fullName = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/strong').text
    name = fullName.split(" ", 2)
    firstName = name[0]
    middleName = name[1]
    lastName = name[2]
    summary = fullName + " is " + arrestWarrant + " and a bond is set for " + bond
    if fullName:
        data_dict['fullName'] = fullName
    if firstName:
        data_dict['firstName'] = firstName
    if middleName:
        data_dict['middleName'] = middleName
    if lastName:
        data_dict['lastName'] = lastName
#     if charges:
#         data_dict['charges'] = charges
    if arrestWarrant:
        data_dict['arrestWarrant'] = arrestWarrant
    if image:
        data_dict['image'] = image
    if gender:
        data_dict['gender'] = gender
    if race:
        data_dict['race'] = race
    if importantDates:
        data_dict['importantDates'] = importantDates
    if additionalInfo:
        data_dict['additionalInfo'] = additionalInfo
    if summary:
        data_dict['summary'] = summary
    data_list.append(data_dict)
    return data_list
```

```python
if __name__ == '__main__':
    data_list = get_data("add_slug_name")
    to_json(data_list)
```
