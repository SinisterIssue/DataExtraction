```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://www.jeffersoncountykssheriff.com/roster.php"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    list1 = driver.find_elements(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div[3]/div/div[1]')
    for i in range (1, 2*len(list1), 2):
        
        list2 = driver.find_elements(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div[3]/div[{i}]/div')
        for j in range (1, len(list2)+1):
            data_dict ={}
            driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div[3]/div[{i}]/div[{j}]/div[2]/div[6]/div/em/a').click()
            fullName = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/span').text
            firstName = fullName.split(" ")[0]
            lastName = fullName.split(" ")[1]

            image = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div/div[1]/div/div[1]/img').get_attribute("src")
            div = len(driver.find_elements(By.XPATH, f'//div[1]/div/div[2]/div/div[@class="inmate_profile_data_bold cell"]/span'))
            for k in range (1, div+1):
                try:
                    headings = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div/div/div/div[1]/div/div[2]/div[{k}]/div[1]').text
            #             print(headings)
                    if "Booking #:"in headings:
                        bookId = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{k}]/div[@class="cell inmate_profile_data_content"]').text
                        print(bookId)
                    elif "Age:" in headings:
                        age = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{k}]/div[@class="cell inmate_profile_data_content"]').text
                        print(age)
                    elif "Gender:" in headings:
                        gender = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{k}]/div[@class="cell inmate_profile_data_content"]').text
                        print(gender)
                    elif "Race:" in headings:
                        race = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{k}]/div[@class="cell inmate_profile_data_content"]').text
                        print(race)
                    elif "Arresting Agency:" in headings:
                        agency = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{k}]/div[@class="cell inmate_profile_data_content"]').text
                        print(agency)
                    elif "Booking Date:" in headings:
                        bookDate = driver.find_element(By.XPATH, f'//div[1]/div/div[2]/div[{k}]/div[@class="cell inmate_profile_data_content"]').text
                        try:
                            bookDate_dmy = bookDate.split(" ")[0]
                            d = bookDate_dmy.split("/")[1]
                            m = bookDate_dmy.split("/")[0]
                            y = bookDate_dmy.split("/")[2]
                            bookDate_dmy = d + "/" + m + "/" + y
                        except:
                            try:
                                bookDate_dmy = bookDate.split(" ")[0]
                                d = bookDate_dmy.split("-")[1]
                                m = bookDate_dmy.split("-")[0]
                                y = bookDate_dmy.split("-")[2]
                                bookDate_dmy = d + "/" + m + "/" + y
                            except:
                                pass
                        print(bookDate)
                        

                    charges = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div/div[1]/div/div[2]/div[8]/div').text.replace("\n", "")
                #     print(charges)
                    bond = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div/div[1]/div/div[2]/div[9]/div[2]').text
                #     print(bond)
                    info = driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/section/div[2]/div/div/div[2]/div').text
                    additionalInfo = "Booking #:" + bookId + ", " + "Arresting Agency:" + agency + ", " + "Bond:" + bond + "," + info 
                #     print(additionalInfo)
                    importantDates = "Booking Date:" + bookDate_dmy
                    summary = fullName + " is charged with " + charges + " and a bond is set for " + bond
                except:
                    pass
            driver.back()
            if fullName:
                data_dict['fullName'] = fullName
            if firstName:
                data_dict['firstName'] = firstName
            if lastName:
                data_dict['lastName'] = lastName
            if charges:
                data_dict['charges'] = charges
            if image:
                data_dict['image'] = image
            if gender:
                data_dict['gender'] = gender
            if race:
                data_dict['race'] = race
            if importantDates:
                data_dict['importantDates'] = importantDates
            if additionalInfo:
                data_dict['additionalInfo'] = additionalInfo
            if summary:
                data_dict['summary'] = summary
            data_list.append(data_dict)
    return data_list
```
