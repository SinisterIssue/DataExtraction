```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
from datetime import date, timedelta
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "http://plaquemines.lavns.org/"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    list1 = driver.find_elements(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr')
    for i in range (2, len(list1)+1):
        data_dict = {}
        list2 = driver.find_elements(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr[{i}]/td')
        for j in range (2, len(list2)+1):
            if j==2:
                fullName = driver.find_element(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr[{i}]/td[{j}]').text
                firstName = fullName.split(",", 1)
                lastName = firstName[0]
                firstName = firstName[1]
            elif j==3:
                dob = driver.find_element(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr[{i}]/td[{j}]').text
            elif j==4:
                race = driver.find_element(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr[{i}]/td[{j}]').text
            elif j==5:
                gender = driver.find_element(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr[{i}]/td[{j}]').text
            elif j==6:
                date = driver.find_element(By.XPATH, f'//div/div/table[@id="gvRoster"]/tbody/tr[{i}]/td[{j}]').text
                importantDates = "Arrest Date: " + date
        if fullName:
            data_dict['fullName'] = fullName
        if firstName:
            data_dict['firstName'] = firstName
        if lastName:
            data_dict['lastName'] = lastName
        if dob:
            data_dict['dob'] = dob
        if race:
            data_dict['race'] = race
        if gender:
            data_dict['gender'] = gender
        if len(date.strip()):
            data_dict['importantDates'] = importantDates
        data_list.append(data_dict)

    return data_list
            
        
```

```python
if __name__ == '__main__':
    data_list = get_data("add_slug_name")
    to_json(data_list)
```
