```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
import time
import json
import re
import string
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("RepresentativesOfReligiousGroups.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "http://www.parliament.cy/en/house-of-representatives"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    driver.find_element(By.XPATH, f'//div/div[4]/div/ul/li[2]/a').click()
    time.sleep(1)
    driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[3]/div/div[2]/div[3]/div/div[1]/div/a').click()
    list1 = driver.find_elements(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[3]/div/div[2]/div')
    for i in range(1, len(list1)+1):
        data_dict = {}
        driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[3]/div/div[2]/div[{i}]/div/div[1]/div/a').click()

        name = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[1]/div').text.title()
        firstName = name.split(" ")[0]
        lastName = name.split(" ")[1]
        print(name)

        careerInfoDesignation = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[1]/em').text
        print(careerInfoDesignation)

        image = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/img').get_attribute("src")
        print(image)

        birthInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[9]').text
        birthInfo = birthInfo.split("\n")[1]
        dob = birthInfo.split(",")[-1].strip()
        placeOfBirthCity = birthInfo.split(",")[0]
        print(dob)
        print(placeOfBirthCity)

        maritalStatus = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[10]').text
        maritalStatus = maritalStatus.split("\n")[1]
        print(maritalStatus)

        educationInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[11]').text
        educationInfo = educationInfo.split("\n")[1]
        print(educationInfo)

        careerInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[12]').text
        careerInfo = careerInfo.split("\n")[1]
        print(careerInfo)

        languagesKnown = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[13]').text
        languagesKnown = languagesKnown.split("\n")[1]
        print(languagesKnown)

        info = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[15]').text.replace("\n", " ")
        currentInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[16]').text.replace("\n", " ")

        listOfCareerInfo = info + "; " + currentInfo
        print(listOfCareerInfo)

        politicalPartyCareer = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[17]').text
        if politicalPartyCareer == "Political career:":
            politicalPartyCareer = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[18]').text.replace("-", "").strip()
            politicalPartyCareer = politicalPartyCareer.replace("\u2019", "'").replace("\n", ", ")
            print(politicalPartyCareer)
        else:
            politicalPartyCareer = ""
        if i==1:
            contactInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[19]').text
            fullAddress = contactInfo.split("\n")[0].split(":")[1]
            telephoneNos = contactInfo.split("\n")[1].split(":")[1]
            fax = contactInfo.split("\n")[2].split(":")[1]
            emails = contactInfo.split("\n")[3].split(":")[1]
            website = ""
        if i==2:
            contactInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[20]').text
            fullAddress = contactInfo.split("\n")[0].split(":")[1]
            telephoneNos = contactInfo.split("\n")[1].split(":")[1]
            fax = contactInfo.split("\n")[2].split(":")[1]
            emails = contactInfo.split("\n")[3].split(":")[1]
            website = ""
        if i==3:
            contactInfo = driver.find_element(By.XPATH, f'/html/body/div[1]/div/div/div[2]/section[2]/div/div[2]/div/p[18]').text
            fullAddress = contactInfo.split("\n")[0].split(":")[1]
            telephoneNos = contactInfo.split("\n")[1].split(":")[1]
            fax = contactInfo.split("\n")[2].split(":")[1]
            emails = contactInfo.split("\n")[3].split(":")[1]
            website = contactInfo.split("\n")[4].split(":")[1]
        summary = name + " is the " + careerInfoDesignation + " and " + maritalStatus

        print("*"*50)
        if name:
            data_dict['fullName'] = name
        if firstName:
            data_dict['firstName'] = firstName
        if lastName:
            data_dict['lastName'] = lastName
        if careerInfoDesignation:
            data_dict['careerInfoDesignation'] = careerInfoDesignation
        if image:
            data_dict['image'] = image
        if dob:
            data_dict['dob'] = dob
        if placeOfBirthCity:
            data_dict['placeOfBirthCity'] = placeOfBirthCity
        if maritalStatus:
            data_dict['maritalStatus'] = maritalStatus
        if educationInfo:
            data_dict['educationInfo'] = educationInfo
        if careerInfo:
            data_dict['careerInfo'] = careerInfo
        if languagesKnown:
            data_dict['languagesKnown'] = languagesKnown
        if listOfCareerInfo:
            data_dict['listOfCareerInfo'] = listOfCareerInfo
        if politicalPartyCareer:
            data_dict['politicalPartyCareer'] = politicalPartyCareer
        if fullAddress:
            data_dict['fullAddress'] = fullAddress
        if telephoneNos:
            data_dict['telephoneNos'] = telephoneNos
        if fax:
            data_dict['fax'] = fax
        if emails:
            data_dict['emails'] = emails
        if website:
            data_dict['website'] = website
        if summary:
            data_dict['summary'] = summary
        data_list.append(data_dict)
        driver.back()
    return data_list
```

```python
if __name__ == '__main__':
    data_list = get_data("add_slug_name")
    to_json(data_list)
```
