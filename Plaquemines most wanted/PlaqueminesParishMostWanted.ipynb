```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://www.ppso.net/wanted.php"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)

    list1 = driver.find_elements(By.XPATH, f'/html/body/div[3]/main/div/div/div[3]/div/div[1]/div/div[1]')
    # print(len(list1))
    # list2 = driver.find_elements(By.XPATH, f'/html/body/div[3]/main/div/div/div[3]/div[{i}]/div/div/div[1]')
    # print(len(list2))
    for i in range (1, len(list1)+1):
        list2 = driver.find_elements(By.XPATH, f'/html/body/div[3]/main/div/div/div[3]/div/div[{i}]/div/div[1]')
        for j in range (1, len(list2)+1):
            data_dict ={}
            driver.find_element(By.XPATH, f'/html/body/div[3]/main/div/div/div[3]/div[{j}]/div[{i}]/div/div[1]/a').click()
            time.sleep(3)
            fullName = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[@id="wanted-name"]').text
            firstName = fullName.split(",", 1)
            lastName = firstName[0]
            firstName = firstName[1].strip()
            print(fullName)

            charges = driver.find_element(By.XPATH, f'/html/body/div/div[@id="wanted-charges"]/div/p').text
            print(charges)
            try:
                address = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[1]/div[@class="right-cell"]').text
                address = address.replace("\n", "")
                gender = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[2]/div[@class="right-cell"]').text
                dob = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[3]/div[@class="right-cell"]').text
                try:
                    d = dob.split("/")[1]
                    m = dob.split("/")[0]
                    y = dob.split("/")[2]
                    dob_dmy = d + "/" + m + "/" + y
                except:
                    try:
                        d = dob.split("-")[2]
                        m = dob.split("-")[1]
                        y = dob.split("-")[0]
                        dob_dmy = d + "/" + m + "/" + y
                    except:
                        pass
                hair = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[4]/div[@class="right-cell"]').text
                eyes = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[5]/div[@class="right-cell"]').text
                height = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[6]/div[@class="right-cell"]').text
                height = height.replace('\"', '')
                weight = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[7]/div[@class="right-cell"]').text
                race = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[8]/div[@class="right-cell"]').text
            except:
                try:
                    address = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[1]/div[@class="right-cell"]').text
                    address = address.replace("\n", "")
                    gender = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[2]/div[@class="right-cell"]').text
                    dob = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[3]/div[@class="right-cell"]').text
                    try:
                        d = dob.split("/")[1]
                        m = dob.split("/")[0]
                        y = dob.split("/")[2]
                        dob_dmy = d + "/" + m + "/" + y
                    except:
                        try:
                            d = dob.split("-")[2]
                            m = dob.split("-")[1]
                            y = dob.split("-")[0]
                            dob_dmy = d + "/" + m + "/" + y
                        except:
                            pass
                    race = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[2]/div[2]/div[4]/div[@class="right-cell"]').text
                    height = ""
                    hair = ""
                    weight = ""
                    eyes = ""
                except Exception:
                    pass
            image = driver.find_element(By.XPATH, f'/html/body/div/div[2]/div[@id="most-wanted-image"]/img').get_attribute("src")
            summary = fullName + " is wanted for " + charges + " last known address was " + address
            driver.back()
            if fullName:
                data_dict['fullName'] = fullName
            if firstName:
                data_dict['firstName'] = firstName
            if lastName:
                data_dict['lastName'] = lastName
            if charges:
                data_dict['charges'] = charges
            if image:
                data_dict['image'] = image
            if address:
                data_dict['listOfAddress'] = address
            if gender:
                data_dict['gender'] = gender
            if dob:
                data_dict['dob'] = dob_dmy
            if hair:
                data_dict['hair'] = hair
            if eyes:
                data_dict['eyes'] = eyes
            if height:
                data_dict['height'] = height
            if weight:
                data_dict['weight'] = weight
            if race:
                data_dict['race'] = race
            if summary:
                data_dict['summary'] = summary
            print(data_dict)
            data_list.append(data_dict)
            
    
    return data_list
```
