```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
import re
from deep_translator import GoogleTranslator
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
def get_data(slug_name):
    data_list = []
    url = "https://www.independentmail.com/story/news/crime/2018/01/13/here-anderson-countys-most-wanted-fugitives/1027910001/"
    options = webdriver.ChromeOptions()
    options.add_argument("--start-maximized") 
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--log-level=3")
    translator = GoogleTranslator(target='english')
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    list1 = driver.find_elements(By.XPATH, f'/html/body/div[2]/main/article/div[4]/h2')
    for i in range(1, len(list1)+1):
        data_dict = {}
        nextName  =""
        fullName = driver.find_element(By.XPATH, f'/html/body/div[2]/main/article/div[4]/h2[{i}]').text
        try:
            driver.find_element(By.XPATH, f'/html/body/div[7]/div/button').click()
        except:
            pass
        try:
            nextName = driver.find_element(By.XPATH, f'/html/body/div[2]/main/article/div[4]/h2[{i+1}]').text
        except:
            pass
        info = driver.find_element(By.XPATH, f'/html/body/div[2]/main/article/div[4]').text
        summary = driver.find_element(By.XPATH, f'/html/body/div[2]/main/article/div[4]/figure[{i}]/div').text
        print(summary)
        try:
            charges = info.split(fullName, 1)[1].split(nextName, 1)[0].strip()
            
        except:
            pass
        image = driver.find_element(By.XPATH, f'/html/body/div[2]/main/article/div[4]/figure[{i}]/img').get_attribute("src")
#         print(image)

        if i == len(list1):
            charges = driver.find_element(By.XPATH, f'/html/body/div[2]/main/article/div[4]/p[15]').text
        print(charges)
        age = fullName.split(",")[1].strip()
        fullName = fullName.split(",")[0].strip()
        print("*"*50)

        if fullName:
            data_dict['fullName'] = fullName
        if age:
            data_dict['age'] = age
        if image:
            data_dict['image'] = image
        if charges:
            data_dict['charges'] = charges
        if summary:
            data_dict['summary'] = summary
        data_list.append(data_dict)
    driver.quit()
    return data_list
```
