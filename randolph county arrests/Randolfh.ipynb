```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
from datetime import date, timedelta
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "https://randolphcounty.us/services/review-recent-arrests"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")
```

```python
def get_data(slug_name):
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    while True:
        try:
            list1 = driver.find_elements(By.XPATH, f'/html/body/div[4]/section/div/div/div/div/div[4]/div/div/div/div[2]/table/tbody/tr')
            for i in range (1, len(list1)+1):
                data_dict = {}
                list2 = driver.find_elements(By.XPATH, f'/html/body/div[4]/section/div/div/div/div/div[4]/div/div/div/div[2]/table/tbody/tr[{i}]/td')
                for j in range (1, len(list2)+1):
                    if j==1:
                        dateArrested = driver.find_element(By.XPATH, f'/html/body/div[4]/section/div/div/div/div/div[4]/div/div/div/div[2]/table/tbody/tr[{i}]/td[{j}]').text
                        d = dateArrested.split("/")[1]
                        m = dateArrested.split("/")[0]
                        y = dateArrested.split("/")[2]
                        dateArrested_dmy = d + "/" + m + "/" + y
                    elif j==2:
                        name = driver.find_element(By.XPATH, f'/html/body/div[4]/section/div/div/div/div/div[4]/div/div/div/div[2]/table/tbody/tr[{i}]/td[{j}]').text
                        fullName = name.split("\n")[0]
                        if "," not in fullName and len(fullName.split(" "))>2:
                            firstName = fullName.split(" ")[0]
                            middleName = fullName.split(" ")[1]
                            lastName = fullName.split(" ")[2]
                        elif "," not in fullName and len(fullName.split(" "))==2:
                            firstName = fullName.split(" ")[0]
                            lastName = fullName.split(" ")[1]
                            middleName = ""
                        elif "," in fullName:
                            firstName = fullName.split(",")[1]
                            lastName = fullName.split(',')[0]
                            middleName = ""
                        arrestingCity = name.split("\n")[1]
                        charges = name.split("\n")[2]
#                         print(fullName)
                    elif j==3:
                        image = driver.find_element(By.XPATH, f'/html/body/div[4]/section/div/div/div/div/div[4]/div/div/div/div[2]/table/tbody/tr[{i}]/td[{j}]').get_attribute("src")
                    elif j==4:
                        agency = driver.find_element(By.XPATH, f'/html/body/div[4]/section/div/div/div/div/div[4]/div/div/div/div[2]/table/tbody/tr[{i}]/td[{j}]').text

                importantDates = "Date Arrested: " + dateArrested_dmy
                additionalInfo = "Arresting City: " + arrestingCity + ", Arresting Agency: " + agency
                summary = fullName + " is charged with " + charges + " and was arrested by " + agency
                if fullName:
                    data_dict['fullName'] = fullName
                if firstName:
                    data_dict['firstName'] = firstName
                if middleName:
                    data_dict['middleName'] = middleName
                if lastName:
                    data_dict['lastName'] = lastName
                if image:
                    data_dict['image'] = image
                if importantDates:
                    data_dict['importantDates'] = importantDates
                if additionalInfo:
                    data_dict['additionalInfo'] = additionalInfo
                if summary:
                    data_dict['summary'] = summary
                data_list.append(data_dict)
            print("1")
            driver.find_element(By.LINK_TEXT, f'next ›').click()
            time.sleep(1)
        except:
            break
    return data_list
    #     time.sleep(4)
```

```python
if __name__ == '__main__':
    data_list = get_data("add_slug_name")
    to_json(data_list)
```
