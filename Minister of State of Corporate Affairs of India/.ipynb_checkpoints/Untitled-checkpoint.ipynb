```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import json
```

```python
def to_json(data):
    """
    This function takes list of dictionary as Input and 
    then Creates a JSON file in which Input data is stored
    """
    with open("data_dict.json", "w") as outfile:
        json.dump(data, outfile,indent=4)
        outfile.close()
```

```python
data_list = []
url = "http://www.mca.gov.in/MinistryV2/minister_ca_state.html"
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized") 
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--log-level=3")

driver = webdriver.Chrome(options=options)
```

```python
driver.get(url)
total_links = driver.find_elements(By.XPATH, f'//div/div/ul/li/a')
links = [i.get_attribute("href") for i in total_links]
for i in links:
    driver.get(i)
    time.sleep(3)
    try:
        name = driver.find_element(By.XPATH, f'//div/div[1]/div[2]//*[@class="profileHeader row"]').text
        name = name.strip()
        print(name)
    except:
#         name = driver.find_element(By.XPATH, f'//div/table/tbody/tr/td[1]').text
#         name = name.strip()
#         print(name)
        pass
    try:
        designation = driver.find_element(By.XPATH, f'//div/div/div[1]/div[2]/p[2]//*[@class="contentdata col-9"]').text
        designation = designation.strip()
        print(designation)
    except:
        try:
            designation = driver.find_element(By.XPATH, f'//div/div/div[2]/p[1]//*[@class="contentdata col-8"]').text
            designation = designation.strip()
            print(designation)
        except:
            pass
    try:
        contactDetails = driver.find_element(By.XPATH, f'//div/div[1]/div[2]/p[3]//*[@class="contentdata col-9"]').text
        contactDetails = contactDetails.strip()
        print(contactDetails)
    except:
        try:
            contactDetails = driver.find_element(By.XPATH, f'//div/div[1]/div[2]/p[3]//*[@class="contentdata col-8"]').text
            contactDetails = contactDetails.strip()
            print(contactDetails)
        except:
            pass
    
    try:
        image = driver.find_element(By.XPATH, f'//div/div/div[1]/div[1]//*[@class="profilecard-img"]').get_attribute("src")
        print(image)
    except:
        try:
            image = driver.find_element(By.XPATH, f'//div/div/div/div[1]//*[@class="profile-pic-image"]').get_attribute("src")
            print(image)
        except:
            pass
    try:
        driver.find_element(By.XPATH, '//div/div/div[1]/div[2]/div//*[@id="show_all_details"]').click()
        time.sleep(3)
    except:
        try:
            driver.find_element(By.XPATH, '//div[3]/div/div/div/div//*[@id="view-ofc"]').click()
            time.sleep(3)
        except:
            pass
        
    
    
```
